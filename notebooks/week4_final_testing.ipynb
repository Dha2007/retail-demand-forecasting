```python
# Test 2: Load all models
print("\n TEST 2: Loading all models...")

with open("../models/lightgbm_model.pkl", "rb") as f:
    basic_model = pickle.load(f)
print("Basic LightGBM loaded!")

with open("../models/advanced_lgbm_model.pkl", "rb") as f:
    advanced_model = pickle.load(f)
print("Advanced LightGBM loaded!")

# Test 3: Model prediction test
print("\n TEST 3: Testing model prediction...")
featured['date'] = pd.to_datetime(featured['date'])
featured['cat_encoded'] = featured['cat_id'].astype('category').cat.codes
featured['store_encoded'] = featured['store_id'].astype('category').cat.codes
featured['state_encoded'] = featured['state_id'].astype('category').cat.codes
featured['dept_encoded'] = featured['dept_id'].astype('category').cat.codes

feature_cols = ['month', 'year', 'wday', 'day', 'week', 'quarter',
                'is_weekend', 'is_month_start', 'is_month_end',
                'has_event', 'is_snap', 'avg_price',
                'sales_lag_1', 'sales_lag_7', 'sales_lag_28',
                'sales_rolling_7', 'sales_rolling_28',
                'cat_encoded', 'store_encoded', 'state_encoded', 'dept_encoded']

sample = featured.head(5)[feature_cols]
predictions = advanced_model.predict(sample)
predictions = np.maximum(predictions, 0)
print("Sample predictions:", predictions.round(2))
print("Model prediction test PASSED!")
```
